```python
cols_str = [col for col in members_df.columns if members_df[col].dtype in ["object", "string"]] 
```

```python
for col in cols_str:
    members_df[col] = members_df[col].astype('str').str.strip()
```

```python
cols_lowercase = ['gender','membership_type' , 'preferred_facility','status']
for col in cols_lowercase:
    members_df[col] = members_df[col].apply(lambda x :x.lower())
```

```python
payments_df = pd.read_csv('data/raw/payments.csv')
```

```python
merged_payments  = payments_df.merge(members_df,on='member_id' , how='inner')
```

```python
merged_payments[['member_id','payment_date', 'amount_pkr','payment_type','payment_status',
    'membership_type','preferred_facility']]
```

```python
merged_payments.groupby(['payment_type','membership_type', 'preferred_facility'])['amount_pkr'].mean()
```

```python
pivot_df = merged_payments.pivot_table(
    values='member_id',   # The data you want to aggregate
    index='preferred_facility',      # Rows of the pivot table
    columns=['membership_type','payment_type'],    # Columns of the pivot table
    aggfunc='count',             # Aggregation (mean, sum, count, max, etc.)
    fill_value=0               # Replace missing values/NaNs
)
pivot_df
```
